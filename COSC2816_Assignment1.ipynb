```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
uci_ml = uci.copy()
```

```python
# Creating binary target
uci_ml["pass"] = (uci_ml["G3"] >= 10).astype(int)
```

```python
# Removing original grades
uci_ml = uci_ml.drop(["G1", "G2", "G3"], axis=1)
```

```python
uci_ml.head()
```

```python
uci_ml.shape
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
# Separating features and target
X_uci = uci_ml.drop("pass", axis=1)
y_uci = uci_ml["pass"]
```

```python
# Finding categorical columns
categorical_cols = X_uci.select_dtypes(include="object").columns
```

```python
# Finding numerical columns
numerical_cols = X_uci.select_dtypes(exclude="object").columns
```

```python
# Encode categorical variables
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ],
    remainder="passthrough"
)

X_uci_encoded = preprocessor.fit_transform(X_uci)
```

```python
X_uci_encoded.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_uci_encoded,
    y_uci,
    test_size=0.2,
    random_state=42,
    stratify=y_uci
)
```

```python
X_uci_encoded.shape
```

```python
#Splitting the UCI data into training and testing sets
from sklearn.model_selection import train_test_split

X_train_uci, X_test_uci, y_train_uci, y_test_uci = train_test_split(
    X_uci_encoded,
    y_uci,
    test_size=0.2,
    random_state=42,
    stratify=y_uci
)
```

```python
print(X_train_uci.shape)
print(X_test_uci.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_uci = RandomForestClassifier(
    random_state=42
)

rf_uci.fit(X_train_uci, y_train_uci)
```

```python
#Making predictions with Random Forest
rf_predictions = rf_uci.predict(X_test_uci)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Random Forest Accuracy:")
print(accuracy_score(y_test_uci, rf_predictions))

print("\nClassification Report:")
print(classification_report(y_test_uci, rf_predictions))
```

```python
#Logitic Regression
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler(with_mean=False)

X_train_scaled = scaler.fit_transform(X_train_uci)
X_test_scaled = scaler.transform(X_test_uci)
```

```python
from sklearn.linear_model import LogisticRegression

lr_uci = LogisticRegression(
    random_state=42,
    max_iter=1000
)

lr_uci.fit(X_train_scaled, y_train_uci)
```

```python
lr_predictions = lr_uci.predict(X_test_scaled)
```

```python
print("Logistic Regression Accuracy:")
print(accuracy_score(y_test_uci, lr_predictions))

print("\nClassification Report:")
print(classification_report(y_test_uci, lr_predictions))
```

```python
import pandas as pd

oulad = pd.read_csv("studentInfo.csv")

print(oulad.shape)
oulad.head()
```

```python
oulad_ml = oulad.copy()

oulad_ml["success"] = oulad_ml["final_result"].apply(
    lambda x: 1 if x in ["Pass", "Distinction"] else 0
)

oulad_ml = oulad_ml.drop(
    ["id_student", "final_result"],
    axis=1
)
```

```python
oulad_ml["success"].value_counts()
```

```python
oulad_features = oulad_ml.drop("success", axis=1)

oulad_encoded = pd.get_dummies(
    oulad_features,
    drop_first=True
)

print(oulad_encoded.shape)
```

```python
oulad_encoded.head()
```

```python
print(oulad_encoded.isnull().sum().sum())
```

```python
print(oulad_encoded.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train_oulad, X_test_oulad, y_train_oulad, y_test_oulad = train_test_split(
    oulad_encoded,
    oulad_ml["success"],
    test_size=0.2,
    random_state=42,
    stratify=oulad_ml["success"]
)

print(X_train_oulad.shape)
print(X_test_oulad.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_oulad = RandomForestClassifier(
    random_state=42
)

rf_oulad.fit(X_train_oulad, y_train_oulad)
```

```python
rf_oulad_predictions = rf_oulad.predict(X_test_oulad)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("OULAD Random Forest Accuracy:")
print(accuracy_score(y_test_oulad, rf_oulad_predictions))

print("\nClassification Report:")
print(classification_report(y_test_oulad, rf_oulad_predictions))
```

```python
from sklearn.preprocessing import StandardScaler

scaler_oulad = StandardScaler(with_mean=False)

X_train_oulad_scaled = scaler_oulad.fit_transform(X_train_oulad)
X_test_oulad_scaled = scaler_oulad.transform(X_test_oulad)
```

```python
from sklearn.linear_model import LogisticRegression

lr_oulad = LogisticRegression(
    random_state=42,
    max_iter=1000
)

lr_oulad.fit(X_train_oulad_scaled, y_train_oulad)
```

```python
lr_oulad_predictions = lr_oulad.predict(X_test_oulad_scaled)
```

```python
print("OULAD Logistic Regression Accuracy:")
print(accuracy_score(y_test_oulad, lr_oulad_predictions))

print("\nClassification Report:")
print(classification_report(y_test_oulad, lr_oulad_predictions))
```

```python
import pandas as pd

rf_importance = pd.DataFrame({
    "feature": X_train_oulad.columns,
    "importance": rf_oulad.feature_importances_
})

rf_importance = rf_importance.sort_values(
    "importance",
    ascending=False
)

rf_importance.head(10)
```

```python
lr_coefficients = pd.DataFrame({
    "feature": X_train_oulad.columns,
    "coefficient": lr_oulad.coef_[0]
})

lr_coefficients["absolute_coefficient"] = (
    lr_coefficients["coefficient"].abs()
)

lr_coefficients = lr_coefficients.sort_values(
    "absolute_coefficient",
    ascending=False
)

lr_coefficients.head(10)
```
